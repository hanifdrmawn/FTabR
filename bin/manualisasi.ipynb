```python
import torch
import torch.nn as nn

# Data input
x_num = torch.tensor([4.1439, 36, 1435, 249, 606, 234, 33.85, -118.15])

# Definisikan parameter
num_features = 8  # Jumlah fitur numerik
embedding_dim = 3  # Dimensi embedding
output_dim = 4  # Dimensi output dari lapisan linear
```

```python
# Definisikan class yang akan mengandung embedding dan lapisan linear
class SimpleModel(nn.Module):
    def __init__(self, num_features, embedding_dim, output_dim):
        super(SimpleModel, self).__init__()
        self.embedding_layer = nn.Embedding(num_features, embedding_dim)
        self.linear = nn.Linear(embedding_dim, output_dim)
    
    def forward(self, x_num_indices):
        # Dapatkan embedding untuk setiap indeks fitur numerik
        x_num_embedded = self.embedding_layer(x_num_indices)
        print("Hasil Embedding untuk tiap fitur numerik:")
        print(x_num_embedded)
        # Terapkan transformasi linear
        x = self.linear(x_num_embedded)
        return x
```

```python
# Inisialisasi model
model = SimpleModel(num_features, embedding_dim, output_dim)
```

```python
# Membuat indeks untuk fitur numerik (0 hingga 7)
x_num_indices = torch.arange(num_features)

# Jalankan forward pass pada model
output = model(x_num_indices)
```

```python
# Print output setelah transformasi linear
print("\nOutput setelah transformasi linear:")
print(output)
```
